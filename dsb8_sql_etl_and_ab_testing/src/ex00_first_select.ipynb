```python
import pandas as pd 
import sqlite3

conn = sqlite3.connect('data/checking-logs.sqlite')
```

```python
schema = pd.io.sql.read_sql("PRAGMA table_info(pageviews);", conn)
```

```python
first_rows = pd.io.sql.read_sql("SELECT * FROM pageviews LIMIT 10;", conn)
print(first_rows)
```

```python
query = """
SELECT uid, datetime 
FROM pageviews 
WHERE uid LIKE 'user_%' 
ORDER BY uid ASC
"""
pageviews = pd.io.sql.read_sql(query, conn, index_col='datetime')
pageviews.index = pd.DatetimeIndex(pageviews.index)
print(pageviews)
```
