```python
import pandas as pd 
import sqlite3

conn = sqlite3.connect('../data/checking-logs.sqlite')
```

```python
schema = pd.io.sql.read_sql("PRAGMA table_info(checker);", conn)
print(schema)
```

```python
first_rows = pd.io.sql.read_sql("SELECT * FROM checker LIMIT 10;", conn)
print(first_rows)
```

```python
query = """
SELECT COUNT(*) as cnt
FROM pageviews p
WHERE EXISTS (
    SELECT 1 
    FROM checker c 
    WHERE c.uid = p.uid 
      AND c.status = 'ready'
      AND c.numTrials = 1
      AND c.labname IN ('laba04', 'laba04s', 'laba05', 'laba06', 'laba06s', 'project1')
)
"""

checkers = pd.io.sql.read_sql(query, conn)
print(checkers)
```
